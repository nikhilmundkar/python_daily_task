```python
import numpy as np
import matplotlib.pyplot as plt

# Data
X = np.linspace(0, 10, 100)

# Create the multiplot figure
fig, axes = plt.subplots(nrows=2, ncols=2, dpi=69, figsize=(12,8))

# Resize figure in inches
fig.set_size_inches(14, 8)

# Change background color
fig.set_facecolor("maroon")

# Add a main title
fig.suptitle("Figure Title", fontsize=25, fontweight="bold")

# Plot 1: sine
axes[0,0].plot(X, np.sin(X), color="blue", linewidth=4)
axes[0,0].set_title("Sine Plot")

# Plot 2: cosine
axes[0,1].plot(X, np.cos(X), color="olive", linewidth=4)
axes[0,1].set_title("Cos Plot")

# Plot 3: tanh
axes[1,0].plot(X, np.tanh(X), color="orange", linewidth=4)
axes[1,0].set_title("Tanh Plot")

# Plot 4: squared
axes[1,1].plot(X, X**2, color="purple", linewidth=4)
axes[1,1].set_title("Squared Plot")

# Show the figure
plt.show()
```
