# 7. Create a bar chart showing the sales of 5 different products.

```python
Patanjali = ["Ghee","Honey","Biscuits","Shampoo","Oil"]
Sales = [120,40,50,100,80]

plt.bar(Patanjali,Sales,color=["green","red","blue","pink","violet"])
plt.xlabel("Patanjali")
plt.ylabel("Sales")
plt.title("Patanjali Product Sales")

plt.grid(axis='y', linestyle='-', alpha=0.7)

plt.show()
```

# 8. Create a horizontal bar chart for the same product-sales data.

```python
Patanjali = ["Ghee","Honey","Biscuits","Shampoo","Oil"]
Sales = [120,40,50,100,80]

plt.barh(Patanjali,Sales,color=["green","red","blue","pink","violet"])
plt.xlabel("Patanjali")
plt.ylabel("Sales")
plt.title("Patanjali Product Sales")

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

# 10.Create a histogram showing the distribution of 100 randomly generated values.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate 100 random values from a normal distribution
data = np.random.randn(100)

# Create histogram
plt.hist(data, bins=10, color='skyblue', edgecolor='black')

# Add labels and title
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of 100 Random Values")

# Show grid
plt.grid(True)

# Display plot
plt.show()
```

# 11.Create a pie chart showing the percentage of customers from 4 different cities.

```python
import matplotlib.pyplot as plt

Customers = [11,12,13,14,15]
Cities = ["Pune","Mumbai","Patna","HYD","Nanded"]

plt.pie(Customers , labels=Cities, autopct="%1.1f%%",startangle=90,colors=["blue","pink","yellow","red"])

plt.title("percentage of customers from 4 cities")

plt.show()
```
